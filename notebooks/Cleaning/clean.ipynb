```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../../data/raw/dataset-6ab989b678d90892980305.csv")
```

We start by dropping the only missing Value we have in this dataset since we have a big dataset and dropping it wont impact our model's training

```python
df = df.dropna()
```

```python
df = df.drop(columns="CUST_ID")
```

Now we save the Dataframe to a csv file for future usage

```python
df.to_csv("../../data/processed/df_clean.csv",index=False)
```
